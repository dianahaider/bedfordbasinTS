### Import and manage BBMP data

#### Metadata __md__ is formatted. It contains 38 columns.
__md__ is the lab's metadata for sampling, extraction and sequencing. \
__dfo_md__ is BBMP remote sensing data (salinity, pH, temperature, density..) \
__bio_niskin__ is nutrient data \
Format __bio_niskin__ data to merge with __md__. __bio_niskin__ is 32 columns, including year, month, day, and depth. __dfo_md__ also has 32 columns, including year_time, month_time, day_time. To merge these data with __md__, we will change the time stamps columns to the same name, and generate a time_string column.

### Detect and plot anomalies in variables

```python
def detect_anomalies(metadata, depth, yr=all, month=all):
    md_col = depth[['weekn', metadata, "year", "month"]].copy()
    md_col = md_col[md_col[metadata].notna()]
    if yr != all:
        #mdcol_yr = md_col[md_col.Year == yr]
        mdcol_yr = md_col[md_col['year'].isin(yr)]
    else: 
        mdcol_yr = md_col
        
    if month != all:
        #mdcol_yr = mdcol_yr[mdcol_yr.Month == month]
        mdcol_yr = mdcol_yr[mdcol_yr['month'].isin(month)]
    
    mdcol_yr = mdcol_yr.drop(columns=['year', "month"])
    mdcol_yr = mdcol_yr.set_index(['weekn'])
    
    #modelling time
    outliers_fraction = float(.01)
    scaler = StandardScaler()
    np_scaled = scaler.fit_transform(mdcol_yr.values.reshape(-1, 1))
    data = pd.DataFrame(np_scaled)
    # train isolation forest
    model =  IsolationForest(contamination=outliers_fraction)
    model.fit(data)
    
    #predict data
    mdcol_yr['anomaly'] = model.predict(data)
    
    
    # visualization
    fig, ax = plt.subplots(figsize=(10,6))
    a = mdcol_yr.loc[mdcol_yr['anomaly'] == -1, [metadata]] #anomaly
    ax.plot(mdcol_yr.index, mdcol_yr[metadata], color='black', label = 'Normal')
    ax.scatter(a.index,a[metadata], color='red', label = 'Anomaly')
    plt.axvline(36, ls='--')
    plt.legend()
    plt.show();
    #add axes names
```

```python
def make_defract(all_md, separated):
    #make sure all size codes are indicated
    all_md["size_code"] = all_md["sampleid"].str.extract(r'[1-9][0-9]?[A-E]([L-S])')
    all_md["size_code"] = all_md["size_code"].fillna('W')

    #only keep values from weeks 1 to 16
    sep_SL = all_md[
        (all_md.size_code != "W") &
        (all_md.size_code != "P")
    ]
    sep_SL = sep_SL.drop(sep_SL[sep_SL.weekn > 16].index)

    #sum [DNA] of small and large size fractions
    sep_SL['[DNAt]'] = sep_SL.groupby(['weekn', 'depth'])['[DNA]ng/uL'].transform('sum')

    #separate small and large size fraction
    sep_S = sep_SL[sep_SL.size_code == 'S']
    sep_L = sep_SL[sep_SL.size_code == 'L']

    #calculate DNA proportion per size fraction
    sep_SL['DNApr'] = sep_SL['[DNA]ng/uL']/sep_SL['[DNAt]']

    #merge with separated on common columns to get corresponding rel. abundances
    sep_SL = sep_SL[['sampleid', 'DNApr', '[DNAt]']].copy()
    sepSLRA = pd.merge(separated, sep_SL, on=['sampleid'], how='left') #all_md is the metadata file

    #exclude ASVs from the whole water
    sep_SLRA = sepSLRA[
        (sepSLRA.size_code != 'W') &
        (sepSLRA.size_code != 'P')
    ]

    #calculate corrected per sample ratio, and corrected feature frequency of de-fractionated samples
    sep_SLRA['Newfeature_frequency'] = sep_SLRA['feature_frequency'] * sep_SLRA['DNApr']
    sep_SLRA['Newff'] = sep_SLRA.groupby(['feature_id', 'weekn', 'depth'])['Newfeature_frequency'].transform('sum')


    #sep_SLRA = sep_SLRA.drop(['sampleid', 'size_code'], axis=1)
    sep_SLRA['sampleid'] = "BB22." + sep_SLRA['weekn'].astype(str) + sep_SLRA['depth_code'] + "SL"

    #uncomment the line below if keeping small and large original sample
    #sep_SLRA['size_code'] = sep_SLRA['size_code'] + '-DFr'

    #uncomment the line above if merging smallandlarge
    sep_SLRA['size_code'] = 'SL'

    #drop unecessary columns which might rise merging conflicts
    sep_SLRA = sep_SLRA.drop(['feature_frequency', 'Total', 'ratio', 'nASVs', 'weekdepth', 'avg',
                              'diff', 'extraction_date', '[DNA]ng/uL', 'A260/280', 'A260/230',
                              'Newfeature_frequency'], axis=1)
    sep_SLRA.rename(columns={'Newff':'feature_frequency'}, inplace=True)
    sep_SLRA = sep_SLRA.drop_duplicates()

    #recalculate ratios
    sep_SLRA['Total'] = sep_SLRA['feature_frequency'].groupby(sep_SLRA['sampleid']).transform('sum')
    sep_SLRA['ratio'] = sep_SLRA['feature_frequency']/sep_SLRA['Total']
    sep_SLRA['nASVs'] = sep_SLRA['feature_id'].groupby(sep_SLRA['sampleid']).transform('nunique')

    sep_SLRA = sep_SLRA.drop_duplicates()

    #make new df dependingg on plotting needs
    sep_WO = separated[separated.size_code == "W"]
    sep_WO = sep_WO.drop_duplicates()

    sep_PO = separated[separated.size_code == "P"]
    sep_PO = sep_PO.drop_duplicates()

    sep_S = separated[separated.size_code == "S"]
    sep_L = separated[separated.size_code == "L"]


    sep_WO.reset_index(inplace=True, drop=True)
    sep_SLRA.reset_index(inplace=True, drop=True)

    #newseparated = pd.concat([sep_SLRA.reset_index(drop=True), sep_WO.reset_index(drop=True)], axis=0).reset_index(drop=True)
    newseparated = pd.concat([sep_SLRA, sep_WO, sep_PO, sep_L, sep_S], ignore_index=True)

    newseparated['weekdepth'] = newseparated["weekn"].astype(str) + newseparated["depth"].astype(str)
    newseparated['avg'] = newseparated['nASVs'].groupby(newseparated['weekdepth']).transform('mean')
    newseparated['diff'] = newseparated['nASVs'] - newseparated['avg']

    newseparated["rank"] = newseparated.groupby("sampleid")["ratio"].rank(method="average", ascending=False)
    newseparated["ranktot"] = newseparated['rank'] / newseparated['nASVs']

    #calculate shannon diversity index
    grouped = newseparated.groupby('sampleid')['feature_frequency'].apply(list)
    diversity = grouped.apply(shannon)
    newseparated['shannon_diversity'] = newseparated['sampleid'].map(diversity)

    #calculate dnaconc of SL (debugged by chatGPT)
    cleaned_data = newseparated.drop_duplicates(subset=["weekn", "depth", "date", "size_code", "[DNA]ng/uL"])

    sl_fill_values = (
        cleaned_data[cleaned_data["size_code"].isin(["S", "L"])]
        .groupby(["weekn", "depth", "date"], as_index=False)["[DNA]ng/ul"]
        .sum()
    )
    sl_fill_values["size_code"] = "SL"
    newseparated.loc[
        (newseparated["size_code"] == "SL") & (newseparated["[DNA]ng/uL"].isna()),
        "[DNA]ng/uL",] = newseparated.merge(
        sl_fill_values,
        on=["weekn", "depth", "date", "size_code"],
        how="left")["[DNA]ng/uL_y"]

    return newseparated
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def taxonomic_barplots_current(comm, table, depth_value, level, n=10, include_other=True):
    """
    Create stacked bar plots showing taxonomic relative abundances (the top n most abundant taxa)
    for samples collected in August, September, and October (across all years) at a specified depth.
    The x-axis will show "year-month" labels constructed from existing 'year' and 'month' columns.
    For each size code, a stacked bar plot is produced.
    
    Parameters:
      comm          : (str) community name. If comm is 'chloroplast', the taxonomic level is set to 'PRSpecies'
      table         : pandas DataFrame containing at least these columns:
                      - 'depth' (numeric)
                      - 'year' (numeric or string, representing the sample year)
                      - 'month' (numeric, representing the month)
                      - 'date' (optionally, if you still have it, but not used for month filtering here)
                      - the taxonomic column (given by the parameter 'level')
                      - 'feature_id' (used to replace ambiguous taxon names)
                      - 'ratio' (relative abundance)
                      - 'size_code' (a categorical variable)
      depth_value   : (numeric) the depth at which to subset the data
      level         : (str) the taxonomic column to analyze (e.g., 'Genus')
      n             : (int) number of top taxa to include (if include_other=True, top n-1 are shown with the rest combined as "Other")
      include_other : (bool) if True, taxa not in the top list are grouped into a category called "Other"
    """
    # If the community is chloroplast, adjust the taxonomic level.
    if comm == 'chloroplast':
        level = 'PRSpecies'
    
    # 1. Filter data for the specified depth.
    sfd_all = table[table['depth'] == depth_value].copy()
    
    # Use the 'month' column directly for filtering to August (8), September (9), and October (10).
    sfd_all = sfd_all[sfd_all['month'].isin([8, 9, 10])]
    
    if sfd_all.empty:
        print("No data available for the specified depth and months.")
        return
    
    # Construct a "year-month" label using the 'year' and 'month' columns.
    # (This assumes that table has a 'year' column; adjust as needed.)
    sfd_all['year_month'] = sfd_all['year'].astype(str) + '-' + sfd_all['month'].apply(lambda m: f"{int(m):02d}")
    
    # 2. Replace ambiguous taxa labels with the feature_id.
    sfd_all[level] = sfd_all.apply(
        lambda row: row['feature_id'] if row[level] in ['Unassigned', 'uncultured+bacterium', 'c__uncultured', 'g__uncultured']
                    else row[level],
        axis=1
    )
    
    # 3. Determine the top taxa based on global abundance (summing 'ratio').
    global_abundance = sfd_all.groupby(level)['ratio'].sum().sort_values(ascending=False)
    if include_other:
        top_taxa = global_abundance.head(n - 1).index.tolist()
        sfd_all['plot_taxa'] = sfd_all[level].apply(lambda x: x if x in top_taxa else 'Other')
        unique_taxa = ['Other'] + top_taxa
    else:
        top_taxa = global_abundance.head(n).index.tolist()
        sfd_all = sfd_all[sfd_all[level].isin(top_taxa)]
        sfd_all['plot_taxa'] = sfd_all[level]
        unique_taxa = top_taxa
    
    # 4. Create a discrete color palette for the taxa.
    color_palette = plt.get_cmap('tab20').colors  # tab20 provides up to 20 colors.
    color_dict = {taxon: color_palette[i % len(color_palette)] for i, taxon in enumerate(unique_taxa)}
    
    # 5. Loop over each size code to produce a stacked bar plot.
    size_codes = sfd_all['size_code'].unique()
    for size_code in size_codes:
        sfd = sfd_all[sfd_all['size_code'] == size_code].copy()
        
        # Group by 'year_month' and 'plot_taxa' to sum the relative abundance.
        agg_df = sfd.groupby(['year_month', 'plot_taxa'])['ratio'].sum().reset_index()
        
        # Pivot the DataFrame so that each row corresponds to a 'year_month'
        # and columns correspond to the different taxa.
        pivot_df = agg_df.pivot(index='year_month', columns='plot_taxa', values='ratio').fillna(0)
        pivot_df = pivot_df.sort_index()  # sort by year-month
        
        # Create x positions for the bars.
        x = np.arange(len(pivot_df.index))
        time_labels = pivot_df.index.tolist()
        
        # Initialize the plot.
        fig, ax = plt.subplots(figsize=(15, 7))
        bottom_stack = np.zeros(len(x))
        
        # Plot each taxon segment.
        for taxon in pivot_df.columns:
            heights = pivot_df[taxon].values
            ax.bar(x, heights, bottom=bottom_stack,
                   color=color_dict.get(taxon, 'gray'),
                   alpha=0.8,
                   label=taxon)
            bottom_stack += heights
        
        # Customize the plot.
        ax.set_xticks(x)
        ax.set_xticklabels(time_labels, rotation=45, fontsize=12)
        ax.set_ylabel("Relative Abundance", fontsize=14)
        ax.set_xlabel("Year-Month", fontsize=14)
        ax.tick_params(axis='y', labelsize=12)
        ax.set_title(f"Stacked Bar Plot for Size Code: {size_code}, Depth: {depth_value}", fontsize=16)
        
        plt.tight_layout()
        plt.savefig(f'outputs/stacked_bar_plot_size_code_{size_code}_depth_{depth_value}_{include_other}.png',
                    dpi=300, bbox_inches='tight')
        plt.show()
    
    # 6. Create a global legend figure for all taxa.
    fig_legend = plt.figure(figsize=(10, 1))
    legend_handles = [
        plt.Line2D([0], [0], color=color_dict[taxon], lw=10, label=taxon.split("__")[-1])
        for taxon in unique_taxa
    ]
    fig_legend.legend(handles=legend_handles, loc="center", ncol=len(unique_taxa), title=level)
    plt.tight_layout()
    plt.savefig(f"outputs/legend_for_all_depths.png", dpi=300, bbox_inches='tight')
    plt.show()

# Example usage:
# taxonomic_barplots_current('my_comm', current_data, depth_value=10, level='Genus', n=10, include_other=True)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def taxonomic_barplots_weekly(comm, table, depth_value, level, n=10, include_other=True):
    """
    Create stacked bar plots showing the weekly taxonomic relative abundances for samples 
    collected in August, September, and October (across all years) at a specified depth.
    
    Each bar corresponds to one week (as indicated by the 'weekn' column) for a given year.
    On the x-axis, bars are grouped by week number; within each group, a minor tick label shows the year.
    
    The y-axis shows the relative abundance (from the 'ratio' column).
    
    Parameters:
      comm          : (str) community name. If comm=='chloroplast', the taxonomic level is set to 'PRSpecies'
      table         : pandas DataFrame that must include at least the following columns:
                      - 'depth' (numeric)
                      - 'month' (numeric; e.g. 8,9,10)
                      - 'weekn' (numeric; week number)
                      - 'year' (numeric or string)
                      - a taxonomic column (specified by `level`, e.g., 'Genus')
                      - 'feature_id' (for replacing ambiguous taxon labels)
                      - 'ratio' (relative abundance)
                      - 'size_code' (categorical; not used for grouping in this function but can be filtered before calling)
      depth_value   : (numeric) the depth at which to subset the data.
      level         : (str) the taxonomic column to analyze (e.g., 'Genus')
      n             : (int) number of top taxa to include. If include_other is True, the top n-1 taxa are shown plus an "Other" category.
      include_other : (bool) if True, taxa not in the top list are grouped into a category called "Other"
    """
    # (Optional) For chloroplast, adjust the taxonomic level.
    if comm == 'chloroplast':
        level = 'PRSpecies'
    
    # 1. Filter by depth and by month using the provided "month" column.
    df = table[table['depth'] == depth_value].copy()
    df = df[df['month'].isin([8, 9, 10])]
    if df.empty:
        print("No data available for the specified depth and months.")
        return
    
    # 2. Replace ambiguous taxon labels with the feature_id.
    df[level] = df.apply(
        lambda row: row['feature_id'] if row[level] in 
            ['Unassigned', 'uncultured+bacterium', 'c__uncultured', 'g__uncultured']
            else row[level],
        axis=1
    )
    
    # 3. Determine the top taxa by summing 'ratio' globally.
    global_abundance = df.groupby(level)['ratio'].sum().sort_values(ascending=False)
    if include_other:
        top_taxa = global_abundance.head(n - 1).index.tolist()
        df['plot_taxa'] = df[level].apply(lambda x: x if x in top_taxa else 'Other')
        unique_taxa = ['Other'] + top_taxa
    else:
        top_taxa = global_abundance.head(n).index.tolist()
        df = df[df[level].isin(top_taxa)]
        df['plot_taxa'] = df[level]
        unique_taxa = top_taxa
    
    # 4. Set up a discrete color palette.
    color_palette = plt.get_cmap('tab20').colors  # up to 20 colors
    color_dict = {taxon: color_palette[i % len(color_palette)] for i, taxon in enumerate(unique_taxa)}
    
    # 5. Aggregate data so that each bar is one (weekn, year) combination.
    # Sum the ratios by week, year, and taxon.
    agg_df = df.groupby(['weekn', 'year', 'plot_taxa'])['ratio'].sum().reset_index()
    
    # Pivot the data so that each row is a unique (weekn, year) and columns are taxa.
    pivot_df = agg_df.pivot_table(index=['weekn','year'], columns='plot_taxa', values='ratio', fill_value=0)
    # Sort first by weekn then by year.
    pivot_df = pivot_df.sort_index(level=['weekn', 'year'])
    
    # 6. Build custom x-axis positions.
    # We want each row (one week for a given year) to have an x position.
    # Also, we want to group by week number.
    x = np.arange(len(pivot_df.index))
    
    # The index is a MultiIndex (weekn, year). Extract labels:
    # For each bar, we want a minor tick showing the year.
    year_labels = [str(yr) for (wk, yr) in pivot_df.index]
    # For major ticks, group by weekn: find the midpoint of x positions for each week.
    groups = {}
    for i, (wk, yr) in enumerate(pivot_df.index):
        groups.setdefault(wk, []).append(i)
    week_midpoints = {wk: np.mean(positions) for wk, positions in groups.items()}
    # Sort week numbers and create labels (casting week numbers to int if necessary)
    unique_weeks = sorted(pivot_df.index.get_level_values('weekn').unique())
    major_positions = [week_midpoints[wk] for wk in unique_weeks]
    week_labels = [f"Week {int(wk)}" for wk in unique_weeks]
    
    # 7. Plot the stacked bar chart.
    fig, ax = plt.subplots(figsize=(15, 7))
    bottom_stack = np.zeros(len(x))
    for taxon in pivot_df.columns:
        heights = pivot_df[taxon].values
        ax.bar(x, heights, bottom=bottom_stack,
               color=color_dict.get(taxon, 'gray'),
               alpha=0.8,
               label=taxon)
        bottom_stack += heights
    
    # Set minor ticks for each bar (labeled with the year).
    ax.set_xticks(x, minor=True)
    ax.set_xticklabels(year_labels, minor=True, rotation=45, fontsize=10)
    
    # Set major ticks at the group midpoints and label with the week number.
    ax.set_xticks(major_positions)
    ax.set_xticklabels(week_labels, fontsize=12)
    
    ax.set_xlabel("Week (grouped by week number; minor tick = Year)", fontsize=14)
    ax.set_ylabel("Relative Abundance (ratio)", fontsize=14)
    ax.set_title(f"Weekly Stacked Bar Plot for Depth {depth_value}", fontsize=16)
    ax.legend(title=level, bbox_to_anchor=(1.05, 1), loc='upper left')
    
    plt.tight_layout()
    plt.show()

# Example usage:
# taxonomic_barplots_weekly('my_comm', your_dataframe, depth_value=10, level='Genus', n=10, include_other=True)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def taxonomic_barplots_weekly(comm, table, depth_value, level, n=10, include_other=True):
    """
    Create stacked bar plots showing weekly taxonomic relative abundances for samples 
    (from months 8, 9, and 10) at a specified depth. Each bar is one (weekn, year) combination.
    
    For the x-axis:
      - Minor tick labels (with pad=5) show the year for each bar.
      - Major tick labels (with pad=40) show the week number (with a gap between week groups).
    
    The y-axis is labeled "Relative Abundance (ratio)".
    
    A separate plot is generated for each unique size code.
    
    Parameters:
      comm          : (str) Community name. If comm=='chloroplast', the taxonomic level is set to 'PRSpecies'
      table         : DataFrame with at least the following columns:
                      'depth', 'month' (numeric), 'weekn' (numeric), 'year', 
                      a taxonomic column (specified by `level` e.g., 'Genus'),
                      'feature_id', 'ratio', and 'size_code'
      depth_value   : numeric; the depth to filter on
      level         : string; the taxonomic column to use (e.g. 'Genus')
      n             : int; number of top taxa to include. If include_other is True, top (n-1) taxa + "Other" are used.
      include_other : bool; if True, taxa not in the top list are grouped into "Other"
    """
    # For chloroplast, adjust the taxonomic level.
    if comm == 'chloroplast':
        level = 'PRSpecies'
    
    # 1. Filter by depth and month (using the "month" column).
    df = table[table['depth'] == depth_value].copy()
    df = df[df['month'].isin([8, 9, 10])].copy()
    if df.empty:
        print("No data available for the specified depth and months.")
        return
    
    # Convert the 'year' column to numeric and then to int (so that sorting is numerical)
    df['year'] = pd.to_numeric(df['year'], errors='coerce')
    df = df.dropna(subset=['year'])
    df['year'] = df['year'].astype(int)
    
    # 2. Replace ambiguous taxon labels with the feature_id.
    ambiguous = ['Unassigned', 'uncultured+bacterium', 'c__uncultured', 'g__uncultured']
    df[level] = df.apply(lambda row: row['feature_id'] if row[level] in ambiguous else row[level], axis=1)
    
    # 3. Determine the top taxa by global abundance.
    global_abundance = df.groupby(level)['ratio'].sum().sort_values(ascending=False)
    if include_other:
        top_taxa = global_abundance.head(n - 1).index.tolist()
        df['plot_taxa'] = df[level].apply(lambda x: x if x in top_taxa else 'Other')
        unique_taxa = ['Other'] + top_taxa
    else:
        top_taxa = global_abundance.head(n).index.tolist()
        df = df[df[level].isin(top_taxa)]
        df['plot_taxa'] = df[level]
        unique_taxa = top_taxa
    
    # 4. Create a discrete color palette.
    color_palette = plt.get_cmap('tab20').colors  # up to 20 colors
    color_dict = {taxon: color_palette[i % len(color_palette)] for i, taxon in enumerate(unique_taxa)}
    
    # 5. Loop over each size code to generate individual plots.
    size_codes = df['size_code'].unique()
    for size_code in size_codes:
        df_size = df[df['size_code'] == size_code].copy()
        
        # Aggregate data by weekn, year, and taxon.
        agg_df = df_size.groupby(['weekn', 'year', 'plot_taxa'])['ratio'].sum().reset_index()
        
        # Pivot the data so each row is a unique (weekn, year) combination.
        pivot_df = agg_df.pivot_table(index=['weekn', 'year'], columns='plot_taxa', values='ratio', fill_value=0)
        # Sort the pivot table first by weekn then by year (numerically).
        pivot_df = pivot_df.sort_index(level=['weekn', 'year'], ascending=[True, True])
        
        # 6. Create custom x-axis positions with a gap between week groups.
        new_x = []             # New x positions for each row of pivot_df
        group_midpoints = {}   # To record the midpoint of each week group (for major tick labels)
        current_week = None
        group_positions = []   # x positions for the current week group
        offset = 0.0
        gap = 0.5              # Extra gap between week groups; adjust as needed
        for (wk, yr) in pivot_df.index:
            if current_week != wk:
                # When a new week group starts, record the midpoint of the previous group (if any)
                if current_week is not None:
                    group_midpoints[current_week] = np.mean(group_positions)
                    offset += gap  # add a gap between groups
                current_week = wk
                group_positions = []
            new_x.append(offset)
            group_positions.append(offset)
            offset += 1.0
        # Finalize the last week group.
        if current_week is not None:
            group_midpoints[current_week] = np.mean(group_positions)
        new_x = np.array(new_x)
        
        # 7. Create tick labels.
        # Minor tick labels: the year for each bar.
        year_labels = [str(yr) for (wk, yr) in pivot_df.index]
        # Major ticks: one per week group.
        unique_weeks = sorted(pivot_df.index.get_level_values('weekn').unique())
        major_positions = [group_midpoints[wk] for wk in unique_weeks]
        week_labels = [f"Week {int(wk)}" for wk in unique_weeks]
        
        # 8. Plot the stacked bar chart.
        fig, ax = plt.subplots(figsize=(15, 7))
        bottom_stack = np.zeros(len(pivot_df.index))
        for taxon in pivot_df.columns:
            heights = pivot_df[taxon].values
            ax.bar(new_x, heights, bottom=bottom_stack,
                   color=color_dict.get(taxon, 'gray'),
                   alpha=0.8,
                   label=taxon)
            bottom_stack += heights
        
        # Set minor ticks for each bar (year labels).
        ax.set_xticks(new_x, minor=True)
        ax.set_xticklabels(year_labels, minor=True, rotation=45, fontsize=10)
        ax.tick_params(axis='x', which='minor', pad=5)  # minor tick pad = 5
        
        # Set major ticks at group midpoints (week labels).
        ax.set_xticks(major_positions)
        ax.set_xticklabels(week_labels, fontsize=14)
        ax.tick_params(axis='x', which='major', pad=40)  # major tick pad = 40
        
        ax.set_xlabel("Week (grouped by week number; minor tick = Year)", fontsize=14)
        ax.set_ylabel("Relative Abundance (ratio)", fontsize=14)
        ax.set_title(f"Weekly Stacked Bar Plot for Depth {depth_value}, Size Code {size_code}", fontsize=16)
        ax.legend(title=level, bbox_to_anchor=(1.05, 1), loc='upper left')
        
        plt.tight_layout()
        plt.show()
        # Optionally, save the figure using plt.savefig(...)
        
# Example usage:
# taxonomic_barplots_weekly('my_comm', your_dataframe, depth_value=10, level='Genus', n=10, include_other=True)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def taxonomic_barplots_weekly(comm, table, depth_value, level, n=10, include_other=True, group_by='week'):
    """
    Create stacked bar plots showing weekly (or monthly) taxonomic relative abundances for samples 
    (from months 8, 9, and 10) at a specified depth. Each bar represents a unique (group, year)
    combination, where "group" is either the week number (if group_by='week') or the month (if group_by='month').
    
    For the x-axis:
      - Minor tick labels (with pad=5) show the year for each bar.
      - Major tick labels (with pad=40) show the group (either "Week X" or "Month X") and a gap is inserted between groups.
    
    If the 'year' value is missing, it is filled by extracting the last two characters from the 'sampleid' column
    (assuming the sample name ends with the two-digit year, e.g. "BB22" → 2022).
    
    A separate plot is generated for each unique size code.
    
    Parameters:
      comm          : (str) Community name. If comm=='chloroplast', the taxonomic level is set to 'PRSpecies'
      table         : DataFrame with at least these columns:
                      - 'depth'
                      - 'month' (numeric; e.g. 8, 9, 10)
                      - 'weekn' (numeric; week number; used if group_by=='week')
                      - 'year'
                      - 'sampleid'
                      - a taxonomic column (specified by `level`, e.g. 'Genus')
                      - 'feature_id'
                      - 'ratio' (relative abundance)
                      - 'size_code'
      depth_value   : numeric; the depth to filter on.
      level         : string; the taxonomic column to use (e.g. 'Genus').
      n             : int; number of top taxa to include. If include_other is True, then top (n-1) taxa are used plus an "Other" category.
      include_other : bool; if True, taxa not in the top list are grouped into "Other".
      group_by      : str; either "week" (default) or "month". Determines how the x-axis major ticks are grouped.
    """
    # Adjust the taxonomic level if needed.
    if comm == 'chloroplast':
        level = 'PRSpecies'
    
    # 1. Filter by depth and month.
    df = table[table['depth'] == depth_value].copy()
    df = df[df['month'].isin([8, 9, 10])].copy()
    if df.empty:
        print("No data available for the specified depth and months.")
        return
    
    # 1a. Fill missing 'year' values by extracting from 'sampleid'
    # Assumes sampleid format like "BB22" (year = 2000 + 22 → 2022)
    if 'year' in df.columns:
        missing_years = df['year'].isnull()
        if missing_years.any():
            if 'sampleid' in df.columns:
                df.loc[missing_years, 'year'] = df.loc[missing_years, 'sampleid'].apply(
                    lambda s: 2000 + int(s[-2:]) if (isinstance(s, str) and len(s) >= 2 and s[-2:].isdigit())
                    else np.nan
                )
            else:
                print("Year is missing and no 'sampleid' column is available to extract it.")
    else:
        if 'sampleid' in df.columns:
            df['year'] = df['sampleid'].apply(
                lambda s: 2000 + int(s[-2:]) if (isinstance(s, str) and len(s) >= 2 and s[-2:].isdigit())
                else np.nan
            )
    
    df = df.dropna(subset=['year'])
    df['year'] = pd.to_numeric(df['year'], errors='coerce').astype(int)
    
    # 2. Replace ambiguous taxon labels with the feature_id.
    ambiguous = ['Unassigned', 'uncultured+bacterium', 'c__uncultured', 'g__uncultured']
    df[level] = df.apply(lambda row: row['feature_id'] if row[level] in ambiguous else row[level], axis=1)
    
    # 3. Determine the top taxa by global abundance.
    global_abundance = df.groupby(level)['ratio'].sum().sort_values(ascending=False)
    if include_other:
        top_taxa = global_abundance.head(n - 1).index.tolist()
        df['plot_taxa'] = df[level].apply(lambda x: x if x in top_taxa else 'Other')
        unique_taxa = ['Other'] + top_taxa
    else:
        top_taxa = global_abundance.head(n).index.tolist()
        df = df[df[level].isin(top_taxa)]
        df['plot_taxa'] = df[level]
        unique_taxa = top_taxa
    
    # 4. Create a discrete color palette.
    color_palette = plt.get_cmap('tab20').colors
    color_dict = {taxon: color_palette[i % len(color_palette)] for i, taxon in enumerate(unique_taxa)}
    
    # 5. Loop over each size code to generate individual plots.
    size_codes = df['size_code'].unique()
    for size_code in size_codes:
        df_size = df[df['size_code'] == size_code].copy()
        
        # 5a. Aggregate the data.
        # Depending on group_by, use either 'weekn' or 'month' along with 'year'
        if group_by.lower() == 'month':
            agg_df = df_size.groupby(['month', 'year', 'plot_taxa'])['ratio'].sum().reset_index()
            pivot_df = agg_df.pivot_table(index=['month', 'year'], columns='plot_taxa', values='ratio', fill_value=0)
            pivot_df = pivot_df.sort_index(level=['month', 'year'], ascending=[True, True])
            grouping_key = 'month'
            major_label_format = lambda m: f"Month {int(m)}"
        else:
            agg_df = df_size.groupby(['weekn', 'year', 'plot_taxa'])['ratio'].sum().reset_index()
            pivot_df = agg_df.pivot_table(index=['weekn', 'year'], columns='plot_taxa', values='ratio', fill_value=0)
            pivot_df = pivot_df.sort_index(level=['weekn', 'year'], ascending=[True, True])
            grouping_key = 'weekn'
            major_label_format = lambda m: f"Week {int(m)}"
        
        # 6. Create custom x-axis positions with a gap between groups.
        new_x = []             # New x positions for each row in pivot_df.
        group_midpoints = {}   # To record the midpoint of each group (for major tick labels).
        current_group = None
        group_positions = []   # x positions for the current group.
        offset = 0.0
        gap = 0.5              # Extra gap between groups; adjust as needed.
        # The pivot_df.index is a MultiIndex; the first level is the grouping key.
        for (grp, yr) in pivot_df.index:
            if current_group != grp:
                if current_group is not None:
                    group_midpoints[current_group] = np.mean(group_positions)
                    offset += gap  # add gap when group changes.
                current_group = grp
                group_positions = []
            new_x.append(offset)
            group_positions.append(offset)
            offset += 1.0
        if current_group is not None:
            group_midpoints[current_group] = np.mean(group_positions)
        new_x = np.array(new_x)
        
        # 7. Create tick labels.
        # Minor tick labels: the year for each bar.
        year_labels = [str(yr) for (grp, yr) in pivot_df.index]
        # Major ticks: one per group (week or month).
        unique_groups = sorted(pivot_df.index.get_level_values(0).unique())
        major_positions = [group_midpoints[grp] for grp in unique_groups]
        major_labels = [major_label_format(grp) for grp in unique_groups]
        
        # 8. Plot the stacked bar chart.
        fig, ax = plt.subplots(figsize=(15, 7))
        bottom_stack = np.zeros(len(pivot_df.index))
        for taxon in pivot_df.columns:
            heights = pivot_df[taxon].values
            ax.bar(new_x, heights, bottom=bottom_stack,
                   color=color_dict.get(taxon, 'gray'),
                   alpha=0.8,
                   label=taxon)
            bottom_stack += heights
        
        # Set the tick labels.
        if len(new_x) == 1:
            # If only one bar exists, force it to show both major and minor labels.
            ax.set_xticks(new_x)
            ax.set_xticklabels(major_labels, fontsize=14)
            ax.xaxis.set_minor_locator(plt.FixedLocator(new_x))
            ax.xaxis.set_minor_formatter(plt.FixedFormatter(year_labels))
            plt.setp(ax.get_xminorticklabels(), visible=True)
        else:
            # Minor ticks: each bar (year labels).
            ax.set_xticks(new_x, minor=True)
            ax.set_xticklabels(year_labels, minor=True, rotation=45, fontsize=10)
            ax.xaxis.set_minor_formatter(plt.FixedFormatter(year_labels))
            ax.tick_params(axis='x', which='minor', pad=5)
            # Major ticks: one per group.
            ax.set_xticks(major_positions)
            ax.set_xticklabels(major_labels, fontsize=14)
            ax.tick_params(axis='x', which='major', pad=40)
        
        ax.set_xlabel(f"{group_by.title()} (grouped by {group_by}); minor tick = Year", fontsize=14)
        ax.set_ylabel("Relative Abundance (ratio)", fontsize=14)
        ax.set_title(f"Stacked Bar Plot for Depth {depth_value}, Size Code {size_code}", fontsize=16)
        ax.legend(title=level, bbox_to_anchor=(1.05, 1), loc='upper left')
        
        plt.tight_layout()
        plt.show()
        # Optionally, you can save the figure with plt.savefig(...)

# Example usage:
# taxonomic_barplots_weekly('my_comm', your_dataframe, depth_value=10, level='Genus', n=10, include_other=True, group_by='month')
```

```python
import plotly.express as px
from sklearn.decomposition import PCA
from sklearn import datasets
from sklearn.preprocessing import StandardScaler

df = df2
features = ['pH','Temperature', 'temperature', 'Chlorophyll A', 'Nitrate', 'Phosphate', 'Silicate', 'Salinity']
X = df[features]

pca = PCA(n_components=2)
components = pca.fit_transform(X)

loadings = pca.components_.T * np.sqrt(pca.explained_variance_)

fig = px.scatter(components, x=0, y=1, color=df['Year'])

for i, feature in enumerate(features):
    fig.add_annotation(
        ax=0, ay=0,
        axref="x", ayref="y",
        x=loadings[i, 0],
        y=loadings[i, 1],
        showarrow=True,
        arrowsize=2,
        arrowhead=2,
        xanchor="right",
        yanchor="top"
    )
    fig.add_annotation(
        x=loadings[i, 0],
        y=loadings[i, 1],
        ax=0, ay=0,
        xanchor="center",
        yanchor="bottom",
        text=feature,
        yshift=5,
    )
fig.show()
```
