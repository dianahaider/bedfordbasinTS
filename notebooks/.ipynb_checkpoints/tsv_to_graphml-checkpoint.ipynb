```python
#adapted from chatGPT
import pandas as pd
import networkx as nx

# 1) Load correlation matrix (rows & cols = OTU IDs)
cor = pd.read_csv('../sparCC/SparCC3/sparcc_cor_fall.tsv',
                  sep='\t', index_col=0)

# 2) (Optional) load two-sided p-values for filtering
pvals = pd.read_csv('../sparCC/SparCC3/results/pvals/pvals.two_sided.txt',
                    sep='\t', index_col=0)

# 3) Define thresholds
corr_thresh = 0.3
pval_thresh = 0.05

# 4) Build an undirected graph
G = nx.Graph()
# add all OTUs as nodes (so isolates appear too)
G.add_nodes_from(cor.index)

# 5) Add edges for significant correlations
for i in cor.index:
    for j in cor.columns:
        if i < j:
            r = cor.at[i, j]
            p = pvals.at[i, j]
            if abs(r) >= corr_thresh and p <= pval_thresh:
                G.add_edge(i, j, weight=r, pvalue=p)

# 6) Export to GraphML
nx.write_graphml(G, 'sparcc_network.graphml')
```
